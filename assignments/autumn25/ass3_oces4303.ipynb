c) ***Plagiarism***: By all means consult each other, work together, or use AI tools, but the files you hand in should be done and written up separately. For allowing checks in Turnitin, you should pass it through `Kernel -> Restarts & Clear Output` before you hand it in. **The default for anyone accused with plagiarism is ZERO on the assignment**, and depending on whether you decide to contest and the result of the appeal, possibly lead to an official note of plagiarism on your transcript (I will allow people to argue but be be ready for possible consequences). 

A few things count as plagiarism:

**Copying between students, and the default is that ALL parties involved get zero for the assignment**, regardless of whether the side can demonstrate they were copied from (extra incentive to keep the writing separate).

**Copying text without citation is plagiarism**. Use quotation marks and give reference if you are directly lifting text, but don't do this too often (and will result in text looking cluttered, and not getting full credit for the *presentation* aspect). This includes content generated from AI.

**Code is a slightly more grey area**, but I will just say no one has ever really been punished for being cautious and generous with citations, but make sure you present it well (e.g. overburdening text with citations will make the presentation ugly, and will not get full credit for the *presentation* aspect say). This includes content generated from AI.

I will just make the point that we don't tend to accuse plagiarism unless we have enough proof, and if we are doing it it probably means we think we have a sufficiently strong case that is probably not worth arguing against (because then penalty then gets increased).

# Assessment 3 (20% of total course grade)

You guessed it: for this assignment you have a choice of using the Argo data (`argo_data_scarp.ipynb`), the bacteria data (`bacteria_data_scrap.ipynb`), or some model data (e.g. the Lotka-Volterra model or the Lorenz model that was in `10_more_neural_networks`), or some other (not too big data) of your choice, somewhat up to you. Pros and cons are that:

* Argo data has substantially fewer variables, but more effort might be needed to load the data.
* Bacteria data is easy to load but has a ton of variables (biogeochemistry focused).
* Model data is easy to generate but slightly restricts to RNNs, altohugh I would then compare the RNNs with CNNs and ANNs to see what benefit the RNNs give.
* Other data (e.g. turtles/penguins data) is fine, but you need to write enough code to make sure I would be able to run the notebook from fresh (which is a requirement for this course relating to *reproducibility*)

With any of the choices you should demonstrate the ability to use neural networks (in one or multiple forms) to do classification and/or regression tasks; you can get full marks in principle using one type of neural networks to either a classification/regression task but ***if you do it well*** (e.g. exploring sensitivity, robustness/cross-validation, comparison with previous methods, leveraging previous techniques encountered etc.). In any case if you do something you should justify why you are doing it with a scientific/practical reason (and comment accordingly on what happens if you don't do it). 

You should minimally do the following:

a) Describe what the data includes, and explain why one might want to do regression and/or classification with it [*be able to do some background research*]

c) Write some of these things up and describe them using the Markdown cells [*practise and demonstrate understanding of Jupyter notebooks*]
